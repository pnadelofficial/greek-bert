```python
result_files = [
    "morpheus_thuc_eval.json",
    "morpheus_soph_eval.json",
    "morpheus_hero_eval.json",
    "morpheus_thuc_artisto_eval.json",
    "morpheus_soph_artisto_eval.json",
    "morpheus_hero_artisto_eval.json"
]
```

```python
import json

def parse_results(file):
    with open(file) as f:
        data = json.load(f)
    work = data["metadata"]["work_label"]
    morpheus_summary = data["morpheus_metrics"]
    model_summary = data["bert_metrics"]

    if "artisto" in file:
        print(f"Results for {work} (Artisto):")
    else:
        print(f"Results for {work} (Ours):")

    for metric, value in morpheus_summary.items():
        if metric in ["lemma_acc", "postag_acc"]:
            print(f"\tMorpheus - {metric}: {value:.4f}")
    
    for metric, value in model_summary.items():
        if metric in ["lemma_acc", "postag_acc"]:
            print(f"\tModel - {metric}: {value:.4f}")
    
    print()
```

```python
for result_file in result_files:
    parse_results(result_file)
```
